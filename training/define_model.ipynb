## Khai báo đường dẫn và class thay thế

```python
from pathlib import Path
import re
import shutil
import time
import importlib

file_path = Path(
    r"~\miniconda3\envs\ai\Lib\site-packages\ultralytics\nn\modules\block.py"
).expanduser()

backup_path = file_path.with_suffix(file_path.suffix + ".backup")

new_class_code = r"""
class Attention(nn.Module):
    # PSA-compatible linear attention (MobileViTv2 style)

    def __init__(self, dim: int, num_heads: int = 8, attn_ratio: float = 0.5):
        super().__init__()

        self.W_i = nn.Conv2d(dim, 1, 1)
        self.W_k = nn.Conv2d(dim, dim, 1)
        self.W_v = nn.Conv2d(dim, dim, 1)

        self.proj = Conv(dim, dim, 1, act=False)
        self.act = nn.ReLU(inplace=True)

    def forward(self, x):
        B, C, H, W = x.shape

        cs = self.W_i(x)
        cs = cs.view(B, 1, -1)
        cs = torch.softmax(cs, dim=-1)

        k = self.W_k(x).view(B, C, -1)
        v = self.act(self.W_v(x)).view(B, C, -1)

        cv = (k * cs).sum(dim=-1, keepdim=True)

        out = v * cv
        out = out.view(B, C, H, W)

        return self.proj(out)
"""
```

## Hàm thay thế và khôi phục

```python
def backup_file():
    if not file_path.exists():
        raise FileNotFoundError(f"Không tìm thấy file: {file_path}")

    shutil.copy2(file_path, backup_path)
    print(f"Đã backup file gốc tại:\n{backup_path}")


def replace_attention_class():
    content = file_path.read_text(encoding="utf-8")

    pattern = r"class Attention\(.*?\):[\s\S]*?(?=\nclass |\Z)"

    if not re.search(pattern, content):
        raise RuntimeError("Không tìm thấy class Attention để thay thế.")

    new_content = re.sub(
        pattern,
        new_class_code.strip(),
        content,
        count=1
    )

    file_path.write_text(new_content, encoding="utf-8")
    print("Đã thay thế class Attention.")


def restore_file():
    if not backup_path.exists():
        raise FileNotFoundError(f"Không tìm thấy file backup: {backup_path}")

    shutil.copy2(backup_path, file_path)
    print("Đã khôi phục file gốc.")
```
